```python
#### prompttemplate
### in our previous example, we used a simple string as the input to the model. However, in many cases, we want to create more complex prompts that include variables or placeholders. This is where PromptTemplate comes in.
### now we want to standardize the way we create prompts for our model. Instead of manually constructing strings, we can use PromptTemplate to define a template with placeholders that can be filled in with specific values.
from langchain_core.prompts import PromptTemplate
import os
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint  
llm = HuggingFaceEndpoint(
    repo_id="deepseek-ai/DeepSeek-V4.1-Flash-1017",
    task="text-generation",
    huggingfacehub_api_token=os.getenv("HUG_ACCESS_TOKEN"),
    max_new_tokens=100,
)

model = ChatHuggingFace(llm=llm)
prompt = PromptTemplate.from_template(
    "Explain {topic} in simple terms."
)

# Connect the prompt template to the model
chain = prompt | model

# Pass the topic variable
reply = chain.invoke({"topic": "Agentic AI"})

print(reply.content)
```

```python
import os
from dotenv import load_dotenv
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import SystemMessage, HumanMessage
from huggingface_hub import InferenceClient

load_dotenv(override=True)

token = os.getenv("HUG_ACCESS_TOKEN")
assert token, "Hugging Face token not found"

# Initialize Hugging Face Inference
client = InferenceClient(
    api_key=token
)

# Create the prompt template
prompt = ChatPromptTemplate.from_messages([
    MessagesPlaceholder(variable_name="messages")
])

# Pass system and human messages dynamically
prompt_value = prompt.invoke({
    "messages": [
        SystemMessage(content="You are a helpful Python tutor."),
        HumanMessage(content="What is a variable? Give a Python example.")
    ]
})

# Convert LangChain messages to Hugging Face API format
messages = [
    {
        "role": "assistant" if msg.type == "ai" else msg.type,
        "content": msg.content
    }
    for msg in prompt_value.to_messages()
]

# Call a Hugging Face-hosted model
response = client.chat_completion(
    model="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    messages=messages,
    max_tokens=100
)

print(response.choices[0].message.content)
```
