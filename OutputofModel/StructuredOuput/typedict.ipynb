```python

from typing import TypedDict
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv

load_dotenv()

class Person(TypedDict):
    name: str
    age: int
    city: str

model = ChatOpenAI(model="gpt-6-astra")

structured_model = model.with_structured_output(Person)
### with_structured_output is making sure our output is structured an by passing person class we can specify the specific output we want
result = structured_model.invoke(
    "Generate a fictional person's name, age, and city."
)

print(result)
print(type(result))
## drawback of type dict is we trust the model to give the output
### basic type eonforcing can be performed
```
